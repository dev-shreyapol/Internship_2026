```python
"""
1. Create a class BankAccount with the following:
a) Attributes: account_holder, account_number, and balance.
b) Methods: deposit(amount), withdraw(amount), and get_balance().
c) withdraw() should print a message and do nothing if the amount exceeds the current balance.
d) A __str__ method that returns a readable summary of the account.
Create two BankAccount objects and perform a few deposits and withdrawals on each 
"""
```

```python
class BankAccount:
    # Constructor
    def __init__(self, account_holder, account_number, balance):
        self.account_holder = account_holder
        self.account_number = account_number
        self.balance = balance

    # Deposit money
    def deposit(self, amount):
        self.balance += amount
        print(f"₹{amount} deposited successfully.")

    # Withdraw money
    def withdraw(self, amount):
        if amount > self.balance:
            print("Insufficient balance! Withdrawal failed.")
        else:
            self.balance -= amount
            print(f"₹{amount} withdrawn successfully.")

    # Get current balance
    def get_balance(self):
        return self.balance

    # Display account details
    def __str__(self):
        return (f"Account Holder: {self.account_holder}\n"
                f"Account Number: {self.account_number}\n"
                f"Balance: ₹{self.balance}")


# Create two BankAccount objects
account1 = BankAccount("Shreya", "1001", 5000)
account2 = BankAccount("Rahul", "1002", 8000)

# Perform transactions on Account 1
print("----- Account 1 -----")
print(account1)
account1.deposit(2000)
account1.withdraw(1000)
account1.withdraw(7000)   # Exceeds balance
print("Current Balance:", account1.get_balance())

# Perform transactions on Account 2
print("\n----- Account 2 -----")
print(account2)
account2.deposit(1500)
account2.withdraw(3000)
print("Current Balance:", account2.get_balance())
```

```python
"""
2. Create a class Student with attributes name, roll_no, and marks (a list of 5 subject marks). Add
methods:
a) calculate_percentage() — returns the average of all marks.
b) get_grade() — returns 'A' for >= 80, 'B' for >= 60, 'C' for >= 40, and 'Fail' otherwise.
c) display() — prints all details including percentage and grade.
Create a list of 3 Student objects and call display() on each.
"""
```

```python
class Student:

    
    def __init__(self, name, roll_no, marks):
        self.name = name
        self.roll_no = roll_no
        self.marks = marks

    
    def calculate_percentage(self):
        percentage = sum(self.marks) / len(self.marks)
        return percentage

    
    def get_grade(self):
        percentage = self.calculate_percentage()

        if percentage >= 80:
            return "A"
        elif percentage >= 60:
            return "B"
        elif percentage >= 40:
            return "C"
        else:
            return "Fail"

    
    def display(self):
        print("Student Details")
        print("-----------------------------")
        print("Name       :", self.name)
        print("Roll No    :", self.roll_no)
        print("Marks      :", self.marks)
        print("Percentage :", self.calculate_percentage())
        print("Grade      :", self.get_grade())
        print()


students = [
    Student("Shreya", 101, [85, 90, 88, 92, 80]),
    Student("Rahul", 102, [70, 65, 75, 60, 80]),
    Student("Priya", 103, [45, 50, 40, 42, 48])
]


for student in students:
    student.display()
```

```python
"""
3. Create a class Employee with private attributes __name, __emp_id, and __salary. Add:
a) Getter and setter methods for all three attributes.
b) The salary setter should raise a ValueError if the value is below 10000.
c) A method give_raise(percent) that increases salary by the given percentage.
Demonstrate that the private attributes cannot be accessed directly, and that all access goes
through the methods.
"""
```

```python
class Employee:
    def __init__(self, name, emp_id, salary):
        self.__name = name
        self.__emp_id = emp_id

        if salary < 10000:
            raise ValueError("Salary must be at least 10000.")
        self.__salary = salary


    def get_name(self):
        return self.__name

    def get_emp_id(self):
        return self.__emp_id

    def get_salary(self):
        return self.__salary

    
    def set_name(self, name):
        self.__name = name

    def set_emp_id(self, emp_id):
        self.__emp_id = emp_id

    def set_salary(self, salary):
        if salary < 10000:
            raise ValueError("Salary must be at least 10000.")
        self.__salary = salary

    
    def give_raise(self, percent):
        self.__salary += self.__salary * (percent / 100)


emp = Employee("Shreya", 101, 25000)


print("Name:", emp.get_name())
print("Employee ID:", emp.get_emp_id())
print("Salary:", emp.get_salary())


emp.give_raise(10)
print("Salary after 10% raise:", emp.get_salary())


emp.set_salary(30000)
print("Updated Salary:", emp.get_salary())


try:
    emp.set_salary(8000)
except ValueError as e:
    print("Error:", e)


try:
    print(emp.__salary)
except AttributeError:
    
    print("Cannot access private attribute '__salary' directly.")
print("Access through getter:", emp.get_salary())
```

```python
"""
4. Build a class SecureVault that stores a private __pin and a private __data dictionary. It should
have:
a) A method add_entry(pin, key, value) that adds to __data only if the pin matches.
b) A method get_entry(pin, key) that returns the value only if the pin matches.
c) Any wrong pin attempt should print 'Access Denied' and return nothing
"""
```

```python
class SecureVault:
    def __init__(self, pin):
        self.__pin = pin
        self.__data = {}

    # Method to add data
    def add_entry(self, pin, key, value):
        if pin == self.__pin:
            self.__data[key] = value
            print("Entry added successfully.")
        else:
            print("Access Denied")

    # Method to get data
    def get_entry(self, pin, key):
        if pin == self.__pin:
            return self.__data.get(key, "Key not found")
        else:
            print("Access Denied")
            return None


# Create SecureVault object
vault = SecureVault("1234")

# Add entries with correct PIN
vault.add_entry("1234", "Name", "Shreya")
vault.add_entry("1234", "City", "Pune")

# Retrieve entries with correct PIN
print("Name:", vault.get_entry("1234", "Name"))
print("City:", vault.get_entry("1234", "City"))

# Wrong PIN attempts
vault.add_entry("1111", "Age", 20)
print(vault.get_entry("1111", "Name"))
```

```python
"""
5. Create a base class Shape with a method area() that returns 0 and a method describe() that
prints 'I am a shape'. Create three subclasses:
a) Circle(radius) — overrides area() using pi * r^2.
b) Rectangle(length, width) — overrides area().
c) Triangle(base, height) — overrides area().
Each subclass should also override describe() to print something specific about itself. Create one
object of each and call both methods
"""
```

```python
# Calculator class
class Calculator:

    # Add two or three numbers
    def add(self, a, b, c=0):
        return a + b + c

    # Multiply two or three numbers
    def multiply(self, a, b, c=1):
        return a * b * c


# Vector class
class Vector:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    # Overloading + operator
    def __add__(self, other):
        return Vector(self.x + other.x, self.y + other.y)

    # String representation
    def __str__(self):
        return f"({self.x}, {self.y})"


# Create Calculator object
calc = Calculator()

# Test add()
print("Addition of 2 numbers:", calc.add(10, 20))
print("Addition of 3 numbers:", calc.add(10, 20, 30))

# Test multiply()
print("Multiplication of 2 numbers:", calc.multiply(5, 4))
print("Multiplication of 3 numbers:", calc.multiply(5, 4, 2))

# Create Vector objects
v1 = Vector(2, 3)
v2 = Vector(4, 5)

# Add vectors
v3 = v1 + v2

# Print result
print("Resulting Vector:", v3)
```

```python
"""
9. Perform the following file operations. Use plain text files only.
a) Write a list of 5 student names to a file students.txt, one name per line.
b) Read the file back and print each name with its line number.
c) Append two more names to the same file without overwriting existing content.
d) Read the final file and count the total number of students.
e) Copy the contents of students.txt into a new file students_backup.txt
"""
```

```python
# a) Write 5 student names to students.txt
with open("students.txt", "w") as file:
    file.write("Shreya\n")
    file.write("Rahul\n")
    file.write("Priya\n")
    file.write("Amit\n")
    file.write("Sneha\n")

# b) Read the file and print each name with line number
print("Students with Line Numbers:")
with open("students.txt", "r") as file:
    for line_no, name in enumerate(file, start=1):
        print(f"{line_no}. {name.strip()}")

# c) Append two more names
with open("students.txt", "a") as file:
    file.write("Rohan\n")
    file.write("Neha\n")

# d) Read the final file and count total students
with open("students.txt", "r") as file:
    students = file.readlines()

print("\nTotal Number of Students:", len(students))

# e) Copy contents to students_backup.txt
with open("students.txt", "r") as source:
    content = source.read()

with open("students_backup.txt", "w") as backup:
    backup.write(content)

print("Contents copied successfully to students_backup.txt")
```

```python
"""
10. Build a menu-driven Student Record Manager using OOP and file handling. The app should
have a class StudentManager with the following menu options:
1. Add Student — takes name, roll number, and marks for 3 subjects.
2. View All — displays all stored students with their average marks.
3. Search Student — finds a student by roll number.
4. Delete Student — removes a student by roll number.
5. Save to File — writes all records to students_data.txt.
6. Load from File — reads records back from students_data.txt on startup.
7. Exit
Requirements: use a class with a list as internal storage, handle invalid menu choices gracefully
"""
```

```python
class StudentManager:
    def __init__(self):
        self.students = []
        self.load_from_file()

    # 1. Add Student
    def add_student(self):
        name = input("Enter Name: ")
        roll = input("Enter Roll Number: ")

        m1 = float(input("Enter Marks of Subject 1: "))
        m2 = float(input("Enter Marks of Subject 2: "))
        m3 = float(input("Enter Marks of Subject 3: "))

        student = {
            "name": name,
            "roll": roll,
            "marks": [m1, m2, m3]
        }

        self.students.append(student)
        print("Student Added Successfully!")

    # 2. View All Students
    def view_all(self):
        if not self.students:
            print("No student records found.")
            return

        print("\nStudent Records")
        print("-" * 40)
        for student in self.students:
            avg = sum(student["marks"]) / 3
            print(f"Name : {student['name']}")
            print(f"Roll : {student['roll']}")
            print(f"Marks: {student['marks']}")
            print(f"Average: {avg:.2f}")
            print("-" * 40)

    # 3. Search Student
    def search_student(self):
        roll = input("Enter Roll Number to Search: ")

        for student in self.students:
            if student["roll"] == roll:
                avg = sum(student["marks"]) / 3
                print("\nStudent Found")
                print("Name:", student["name"])
                print("Roll:", student["roll"])
                print("Marks:", student["marks"])
                print("Average:", round(avg, 2))
                return

        print("Student Not Found!")

    # 4. Delete Student
    def delete_student(self):
        roll = input("Enter Roll Number to Delete: ")

        for student in self.students:
            if student["roll"] == roll:
                self.students.remove(student)
                print("Student Deleted Successfully!")
                return

        print("Student Not Found!")

    # 5. Save to File
    def save_to_file(self):
        with open("students_data.txt", "w") as file:
            for student in self.students:
                line = f"{student['name']},{student['roll']},{student['marks'][0]},{student['marks'][1]},{student['marks'][2]}\n"
                file.write(line)

        print("Records Saved Successfully!")

    # 6. Load from File
    def load_from_file(self):
        try:
            with open("students_data.txt", "r") as file:
                for line in file:
                    data = line.strip().split(",")
                    if len(data) == 5:
                        student = {
                            "name": data[0],
                            "roll": data[1],
                            "marks": [float(data[2]), float(data[3]), float(data[4])]
                        }
                        self.students.append(student)
        except FileNotFoundError:
            pass


# Main Program
manager = StudentManager()

while True:
    print("\n===== Student Record Manager =====")
    print("1. Add Student")
    print("2. View All")
    print("3. Search Student")
    print("4. Delete Student")
    print("5. Save to File")
    print("6. Load from File")
    print("7. Exit")

    choice = input("Enter your choice: ")

    if choice == "1":
        manager.add_student()

    elif choice == "2":
        manager.view_all()

    elif choice == "3":
        manager.search_student()

    elif choice == "4":
        manager.delete_student()

    elif choice == "5":
        manager.save_to_file()

    elif choice == "6":
        manager.load_from_file()
        print("Records Loaded Successfully!")

    elif choice == "7":
        manager.save_to_file()
        print("Thank You! Exiting Program...")
        break

    else:
        print("Invalid Choice! Please Try Again.")
```
